```python
import pandas as pd
import numpy as np
```

```python
forex_data = pd.read_csv("Data/forex_price_data.csv")
trade_log = pd.read_csv("Data/trade_log.csv")
mc_scenarios = pd.read_csv("Data/mc_scenarios.csv")
```

```python
print("Missing Values in Forex Data")
print(forex_data.isnull().sum())

print("\nMissing Values in Trade Log")
print(trade_log.isnull().sum())

print("\nMissing Values in Monte Carlo Data")
print(mc_scenarios.isnull().sum())
```

```python
print("Forex Data Types")
print(forex_data.dtypes)

print("\nTrade Log Data Types")
print(trade_log.dtypes)

print("\nMonte Carlo Data Types")
print(mc_scenarios.dtypes)
```

```python
forex_data["DateTime"] = pd.to_datetime(forex_data["DateTime"])

trade_log["EntryDateTime"] = pd.to_datetime(trade_log["EntryDateTime"])

trade_log["ExitDateTime"] = pd.to_datetime(trade_log["ExitDateTime"])
```

```python
print(forex_data.dtypes)

print(trade_log.dtypes)
```

```python
print("Forex Duplicate Rows:", forex_data.duplicated().sum())

print("Trade Log Duplicate Rows:", trade_log.duplicated().sum())

print("Monte Carlo Duplicate Rows:", mc_scenarios.duplicated().sum())
```

```python
forex_data = forex_data.drop_duplicates()

trade_log = trade_log.drop_duplicates()

mc_scenarios = mc_scenarios.drop_duplicates()
```

```python
print(forex_data.shape)

print(trade_log.shape)

print(mc_scenarios.shape)
```

```python
print(forex_data.columns)

print(trade_log.columns)

print(mc_scenarios.columns)
```

```python
forex_data.to_csv("Cleaned_Data/forex_price_data_cleaned.csv", index=False)

trade_log.to_csv("Cleaned_Data/trade_log_cleaned.csv", index=False)

mc_scenarios.to_csv("Cleaned_Data/mc_scenarios_cleaned.csv", index=False)
```
