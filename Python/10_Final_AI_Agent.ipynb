```python
import pandas as pd
import numpy as np
```

```python
forex_data = pd.read_csv("Signal_Data/forex_with_signals.csv")

forex_data.head()
```

```python
print(forex_data.shape)
```

```python
print(forex_data.columns)
```

```python
def trading_ai(signal):
    if signal == "BUY":
        return "🟢 AI Recommendation: BUY"

    elif signal == "SELL":
        return "🔴 AI Recommendation: SELL"

    else:
        return "🟡 AI Recommendation: HOLD"
```

```python
forex_data["AI_Recommendation"] = forex_data["Signal"].apply(trading_ai)

forex_data[
    ["DateTime", "Close", "Signal", "AI_Recommendation"]
].head(20)
```

```python
print(forex_data["AI_Recommendation"].value_counts())
```

```python
import os

os.makedirs("Final_AI_Output", exist_ok=True)

forex_data.to_csv(
    "Final_AI_Output/final_ai_predictions.csv",
    index=False
)
```
