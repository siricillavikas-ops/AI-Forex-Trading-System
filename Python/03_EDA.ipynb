```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
forex_data = pd.read_csv("Cleaned_data/forex_price_data_cleaned.csv")
trade_log = pd.read_csv("Cleaned_data/trade_log_cleaned.csv")
mc_scenarios = pd.read_csv("Cleaned_data/mc_scenarios_cleaned.csv")
```

```python
forex_data["DateTime"] = pd.to_datetime(forex_data["DateTime"])
```

```python
print("Forex Data:", forex_data.shape)
print("Trade Log:", trade_log.shape)
print("Monte Carlo:", mc_scenarios.shape)
```

```python
forex_data.head()
```

```python
plt.figure(figsize=(12,5))

plt.plot(forex_data["DateTime"], forex_data["Close"])

plt.title("EUR/USD Closing Price Over Time")

plt.xlabel("Date")

plt.ylabel("Closing Price")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(12,5))

plt.plot(forex_data["DateTime"], forex_data["Volume"])

plt.title("Trading Volume")

plt.xlabel("Date")

plt.ylabel("Volume")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(12,5))

plt.plot(forex_data["DateTime"], forex_data["Spread"])

plt.title("Spread Over Time")

plt.xlabel("Date")

plt.ylabel("Spread")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```

```python
trade_log["Direction"].value_counts().plot(kind="bar")

plt.title("Trade Direction")

plt.xlabel("Direction")

plt.ylabel("Number of Trades")

plt.show()
```

```python
trade_log["SignalType"].value_counts().plot(kind="bar")

plt.title("Signal Types")

plt.xlabel("Signal")

plt.ylabel("Count")

plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(trade_log["PnL"], bins=20)

plt.title("Profit and Loss Distribution")

plt.xlabel("PnL")

plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(mc_scenarios["PortfolioPnL"], bins=25)

plt.title("Monte Carlo Portfolio Profit/Loss")

plt.xlabel("Portfolio PnL")

plt.ylabel("Frequency")

plt.show()
```

```python
correlation = forex_data.select_dtypes(include=["number"]).corr()

print(correlation)
```
